```python
def create_graph(tx, graph):
    for node in graph.nodes:
        tx.run("CREATE (:Node {name: $name})", name=node)
    for edge in graph.edges:
        tx.run("MATCH (a:Node {name: $source}), (b:Node {name: $target}) "
               "CREATE (a)-[:RELATIONSHIP]->(b)",
               source=edge[0], target=edge[1])
```

```python
with driver.session() as session:
    session.write_transaction(create_graph, graph[0])
```

### 6.2 Neo4j

```python
from py2neo import Graph, Node, Relationship, Subgraph, NodeMatcher
from typing import Any, Dict, List, Tuple


class CommunityNode:
    def __init__(self, name: str, parent_cluster: Any, is_final_cluster: bool):
        self.name = name
        self.parent_cluster = parent_cluster
        self.is_final_cluster = is_final_cluster


class Community:
    def __init__(self, id: int, nodes: List[CommunityNode]):
        self.id = id
        self.nodes = nodes


class CommunityDetectionResult:
    def __init__(self, communities: Dict[int, Dict[int, Community]]):
        self.communities = communities
```

```python
# 连接到 Neo4j 数据库
Neo4jgraph = Graph("bolt://localhost:7687", auth=("neo4j", "1qa2ws#ED"))

# 清除现有数据（可选，根据需要）
Neo4jgraph.delete_all()
```

```python

def store_community_result_in_neo4j(result: CommunityDetectionResult):
    # 定义颜色映射
    color_mapping = {
        0: "red",
        1: "blue"
    }
    node_matcher = NodeMatcher(Neo4jgraph)

    # 存储节点和关系到 Neo4j
    for level, communities in result.communities.items():
        for community in communities.values():
            for node in community.nodes:
                # 创建节点
                neo4j_node = Node("Person", name=node.name, community=community.id)
                Neo4jgraph.create(neo4j_node)

    # 为节点添加颜色属性
    for level, communities in result.communities.items():
        for community in communities.values():
            color = color_mapping.get(community.id)
            for node in community.nodes:
                neo4j_node = node_matcher.match("Person", name=node.name).first()
                if neo4j_node:
                    neo4j_node["color"] = color
                    Neo4jgraph.push(neo4j_node)


def create_relationships(relationships: List[Tuple[str, str, str]]):
    node_matcher = NodeMatcher(Neo4jgraph)
    for source, target, rel_type in relationships:
        source_node = node_matcher.match("Person", name=source).first()
        target_node = node_matcher.match("Person", name=target).first()
        if source_node and target_node:
            rel = Relationship(source_node, rel_type, target_node)
            Neo4jgraph.create(rel)
```

```python

# 存储社区结果
store_community_result_in_neo4j(detection_result)
```

### 6.3连接边

```python
def create_relationships(edges: List[Tuple[str, str]]):
    for source, target in edges:
        source_node = Neo4jgraph.nodes.match("Person", name=source).first()
        target_node = Neo4jgraph.nodes.match("Person", name=target).first()
        if source_node and target_node:
            relationship = Relationship(source_node, "RELATED_TO", target_node)
            Neo4jgraph.create(relationship)
```
